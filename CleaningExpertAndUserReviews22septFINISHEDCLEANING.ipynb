```python
import pandas as pd

df3 = pd.read_csv(
    "UserReviews43DylanDMB.csv",
    encoding="latin1"
    # Use latin1 encoding so pandas can correctly read special characters in the CSV file
)

df4 = pd.read_csv(
    "ExpertReviewsClean43LIWCforDMB.csv",
    encoding="latin1"
    # Use latin1 encoding so pandas can correctly read special characters in the CSV file
)
```

```python
# to see the columns of each files:
print(df3.columns.tolist())
print(df4.columns.tolist())
```

```python
# to see the first 20 rows of title for each file
print(df3["reviewer"].head(20))
print(df4["reviewer"].head(20))
```

```python
# Keep only the columns needed for the analysis and remove all unnecessary columns
# axis=1 tells pandas to remove columns instead of rows
df3 = df3.drop(df3.columns[5:], axis=1) 
df4 = df4.drop(df4.columns[7:], axis=1) 
```
