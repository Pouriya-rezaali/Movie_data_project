```python
# We create a seperate cleaned title column which we can use for matching
# to remove unnecessary punctuations and spaced we add a library called RE
import re
```

```python
# to standardize movie titles and keep missing titles as missing values we create a function: 
def clean_title(title):
    if pd.isna(title):
        return None

#to convert the title to text:
    title = str(title)
#convert all letters to lowercase:
    title = title.lower()
# romove extra spaces from begining and end:
    title = title.strip()
# to remove punctuations and keep letters, numbers, and spaces:
    title = re.sub(r"[^\w\s]", "", title)
# to replace multiple spaces with one space
    title = re.sub(r"\s+", " ", title)

    return title
```

```python
# now we apply the function to our file:
df1["title_clean"] = df1["title"].apply(clean_title)
df2["title_clean"] = df2["title"].apply(clean_title)
```

```python
# to see how many movies have the same movie_key in each file:
keys_file1 = set(df1["movie_key"])
keys_file2 = set(df2["movie_key"])

common_keys = keys_file1.intersection(keys_file2)

print("Number of movie_keys found in both files:", len(common_keys))
print("\nSample matching movie_keys:")
print(list(common_keys)[:20])
```

```python
# to find movie keys that exist in both datasets
common_keys = set(df1["movie_key"]).intersection(set(df2["movie_key"]))

# now we count how many times each movie key appears in each dataset
counts_file1 = df1["movie_key"].value_counts()
counts_file2 = df2["movie_key"].value_counts()

# to count only those with one repeatation
unique_matches = sum(
    (counts_file1.get(key, 0) == 1) and
    (counts_file2.get(key, 0) == 1)
    for key in common_keys
)

# Print the total number of common movie keys
print("Common movie_keys:", len(common_keys))

# Print the number of unique 1-to-1 matches
print("Unique 1-to-1 matches:", unique_matches)

# Find movie keys that appear more than once in either dataset
ambiguous_matches = [
    key for key in common_keys
    if counts_file1.get(key, 0) > 1
    or counts_file2.get(key, 0) > 1
]
# Print the number of ambiguous movie keys
print("\nAmbiguous matching keys:", len(ambiguous_matches))

# Print the first 20 ambiguous movie keys
print("Sample ambiguous keys:")
print(ambiguous_matches[:20])
```

```python
# we loop through each ambiguous movie key we print a separator and the current movie key

for key in ambiguous_matches:
    print(f"\n{'='*60}")
    print(f"MOVIE KEY: {key}")

    # then we show matching rows from FILE 1
    print("\n--- File 1 ---")
    print(
        df1[df1["movie_key"] == key][
            ["title", "release_year", "movie_key"]
        ].to_string(index=False)
    )

    # then FILE 2
    print("\n--- File 2 ---")
    print(
        df2[df2["movie_key"] == key][
            ["title", "release_date", "movie_key"]
        ].to_string(index=False)
    )

```

```python
# Replace problematic titles using the movie slug from The Numbers URLs

import re
from urllib.parse import urlparse

problem_mask = df1["title"].astype(str).str.contains("�", na=False)

def title_from_url(url):
    # Extract the movie slug from the The Numbers URL
    slug = urlparse(url).path.split("/movie/")[-1]

    # Remove country/year information at the end of the slug
    slug = re.sub(r"-\([^)]*\)$", "", slug)

    # Convert hyphens to spaces
    title = slug.replace("-", " ")

    # Remove remaining parenthetical information
    title = re.sub(r"\([^)]*\)", "", title)

    # Clean extra spaces
    title = re.sub(r"\s+", " ", title).strip()

    return title

# Create replacement titles from the URLs
df1.loc[problem_mask, "title"] = (
    df1.loc[problem_mask, "url"].apply(title_from_url)
)

# Check the result
print("Titles replaced:", problem_mask.sum())
print(
    "Titles containing �:",
    df1["title"].astype(str).str.contains("�", na=False).sum()
)
print("\nSample:")
print(
    df1.loc[problem_mask, ["release_year", "title"]]
    .head(20)
    .to_string(index=False)
)
```
